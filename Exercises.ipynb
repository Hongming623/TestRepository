```python
import pulp  
  
# 创建线性规划问题实例，求最大化问题  
MyProbLP = pulp.LpProblem("Exercise", sense=pulp.LpMaximize)  
  
# 利用字典定义决策变量（也可以逐一定义决策变量），这里定义为0-1变量（Binary）  
x = {}  
for i in range(1, 3):  
    x[i] = pulp.LpVariable('x' + str(i), cat='Continuous')  
  
# Function to maximize
MyProbLP += 10 * x[1] + 18 * x[2]  
  
# Conditions 
MyProbLP += (5*x[1] + 2*x[2] <= 170)  # 不等式约束  
MyProbLP += (2*x[1] + 3*x[2] <= 100)  # 等式约束  
MyProbLP += (x[1] + 5*x[2] <= 150)  # 不等式约束  
  
# Solve function 
MyProbLP.solve()  
  
# Prints status 
print("Status:", pulp.LpStatus[MyProbLP.status])  
  
# prints optimal value for each variable  
for v in MyProbLP.variables():  
    print(v.name, "=", v.varValue)  
  
# Prints out maximum or minimum value
print("F(x) = ", pulp.value(MyProbLP.objective))
```

```python
import networkx as nx
G = nx.DiGraph() #Digraph for with direction, Graph() for no specific direction
G.add_weighted_edges_from( 
[(1, 2, 2), (1,4,4), (2,5,2), (2,3,3), (2,6,3), 
 (3,9,5),(4,7,3), (5,6,3), (6,7,1), (6,9,2.5), 
 (7,8,2), (7,9,2), (8,9,4)])
path = nx.dijkstra_path(G, source=1, target=9) # Dijkstra算法求解最短路径 (Path from 1 to 9)
length = nx.dijkstra_path_length(G, source=1, target=9) # Dijkstra算法求解最短 (Distance from 1 to 9)
print("path:", path)
print("length:", length)
```
